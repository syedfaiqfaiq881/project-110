# Data Types 

There are 3 different categories of data types in python 

- Basic Data types (str, int, float, bool)
- Container Data types (list, Set, Tuple, Dictionary)
- User Defined Data Types -----> Will Discuss in OOP (Object-Oriented Programming)

```python
#list Datatype 
name=[3,4.5,"Hello",True]
print(name)
type(name)
```

```python
#tuple Datatype 
name=(3,4.5,"Hello",True)
print(name)
type(name)
```

```python
#set Datatype 
name={3,4.5,"Hello",True}
print(name)
type(name)
```

```python
# Dictionary Datatype ----> (key:value)
dictionary={"name":"Abdullah","Roll No":1020,}
print(dictionary)
type(dictionary)
```

```python
# With user input 
num1=input("Entr number 1 : ")
print(type(num1))
num2=int(input("Entr number 2 : "))
print(type(num2))
result=num1+num2
print(result)
type(result)
```
