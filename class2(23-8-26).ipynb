# Type Conversion

When we want to add one data type to another is called type conversion.

## 2 Type of Type Conversions: 
- 1. Implicit Type conversion : When python automatically convert the data type.

- 2. Explicit Type conversion : When programmer itself change the data type.

```python
#implicit Example

print(9+9.5) # int & float
print(6j + 6+5j) #int & complex
print(6.5 + 4+2j) #float & complex
```

```python
#Explicit 

# float to int
print(int(7.9))

#int to float
print(float(87))

#str to int
print(int("45"))

#int to str
print(str(34))

#float to str
print(str(6.7))

#str to float
print(float("6.7"))

#int to bool
print(bool(0))
print(bool(1))
print(bool(3))

#bool to int
print(int(False))
print(int(True))

#int to complex
print(complex(5))
```

# Operators
Operators are use use to perform operations on variables and values. Python have following operators:

- Arthematic operators
- Comparison operators
- Logical operators
- Assignment operators
- Identity operators
- Membership operators
- Bitwise operators

```python
# Arthematic Operators

x=10
y=2

#addition
print(x + y)

#subtraction
print(x - y)

#multiplication
print(x * y)

#division
print(x/y) #----> #implicit 

# percent
print(x % y) #---> modulus 

# floor division  ----> int 
print(x//y)  

#exponent
print(x**y)
```

```python
# Comparison operators ---> use to compare the values or variables (Boolean (True , False))

x=15
y=10

# grater then, less then
print(x>y)
print(x<y)

# greater then or equal to , less then or equal to
print(x>=y)
print(x<=y)

# equal-equal to , not-equal to
print(x==y)
print(x!=y)
```
