# Operators
Operators are use use to perform operations on variables and values. Python have following operators:

- Arthematic operators
- Comparison operators
- Logical operators
- Assignment operators
- Identity operators
- Membership operators
- Bitwise operators

```python
#logical operators  ---> And , or , not

# And:
#true + true = true
#true + false=false
#fasle + true = false
#fasle + false = false

x=True
y=False

#1. And operator
print(y and y)  

# or :
#true + true = true
#true + false=true
#fasle + true = true
#fasle + false = false

#2. or operator
print (x or y)

# 3. not operator ---> (opposite)
print(not x)
```

```python
a=12 
b=13
c=14

print(a<=b and c>=b) 
print(a>=b or c<=b)  

print(not(c>b and b>a))

print(b<c or a>c)

print(not(a<c))
```

```python
#  Assignment operators  ----> calculation + assign_value to the same variable

a=10
print(a)

# a=a+1 ---> a = 10+1
a+=1 
print(a)

# a=a-1 ---> a = 11-1
a-=1 
print(a)

# a=a*2 --> a=10*2 
a*=2  
print(a)

# a=a**2 ---> a=20**2
a**=2
print(a)

a/=2
print(a)

a//=2
print(a)

a%=2
print(a)
```

```python
# MemberShip operators ----> Shows the presence of a value in a sequence . (True or Flase)

# in , not in 

x= "Isalamabad"

#in
print("I" in x)
print("sal" in x)

#not in
print("m" not in x)

y=[2,3,4,5,7,"Karachi"]
print( 4 in y)
print(6 not in y)
print("a" in y)
print("karachi" not in y)
```

```python
a=(True, 3,5.6,"Pakistan")

print( 4 in a)         # false
print(5.6 in a)        # true 
print("Pak" not in a)  # true
print(True not in a)       
```

```python
# Identitiy operators  ----> use to compare the objects, not if they are equal, but if the are actually the same object with the same memory location
#is , is not

a=3
b=3

print(a is b)
print(a is not b)

x="Hello"
y="World"

print(x is y)
print(x is not y)

x="Hello world"
y="Hello world"


print(x is y)

# use to check the memory location 
print(id(x))
print(id(y))
```

```python
# Bitwise operators ----> use to compare binary values.

# & operator (and)

print(6 & 3)   

# 110  
# 011
#------
# 010
#-----

# | operator (or)
print(6 | 3)

print(10 | 20) 

#  ----> ~ (not operator)
```

```python
# Program to make marksheet

name=input("Enter your Name: ")
roll_no=input("Enter your Roll No: ")
sub_1=float(input("Enter chemistry marks: "))
sub_2=float(input("Enter physics marks: "))
sub_3=float(input("Enter maths marks: "))
sub_4=float(input("Enter english marks: "))
sub_5=float(input("Enter urdu marks: "))

total_marks=500
obtained_marks = sub_1+sub_2+sub_3+sub_4+sub_5
percentage =(obtained_marks*100)/total_marks
grade=None

if percentage>=90 and percentage<=100:
    grade = "A+"
elif percentage>=80 and percentage<90:
    grade = "A"
elif percentage>=70 and percentage<80:
    grade = "A-"
elif percentage>=60 and percentage<70:
    grade = "B"
elif percentage>=50 and percentage<60:
    grade = "C"
elif percentage>=40 and percentage<50:
    grade = "D"
else:
    grade = "F"

print(f"""
                CITY PUBLIC SCHOOL

    Name: {name}
    Roll No: {roll_no}
    Percentage: {percentage}
    Grade: {grade}

""")    
```
